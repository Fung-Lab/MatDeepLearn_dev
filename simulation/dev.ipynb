```python
import numpy as np
import matplotlib.pyplot as plt

fig = plt.figure(figsize=(4, 3))
X = np.linspace(0.1, 6.0, 200)
lj_pot_y = 4 * ((1 / X) ** 12 - (1 / X) ** 6)
morse_pot_y = (np.exp(-2 * (X - 2)) - 2 * np.exp(-1 * (X - 2)))
# plt.plot(X, lj_pot_y, label='Lennard-Jones')
plt.plot(X, morse_pot_y, label='Morse')
plt.legend()
plt.xlim(0, 6)
plt.ylim(-2, 2)
```
